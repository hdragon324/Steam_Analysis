```python
import requests
from bs4 import BeautifulSoup

header = {
    'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/130.0.0.0 Whale/4.29.282.14 Safari/537.36'
    }

# 오픈월드, dlc 포함 x
URL = 'https://store.steampowered.com/search/results/?query&start=0&count=50&dynamic_data=&sort_by=_ASC&supportedlang=koreana&tags=1695&category1=998&snr=1_7_7_7000_7&filter=topsellers&infinite=1'


response = requests.get(URL, headers=header)
json_data = response.json()
soup = BeautifulSoup(json_data['results_html'], 'html.parser')
soup
```

```python
# 게임 원가 추출
price = soup.select('a.search_result_row div.search_discount_and_price')

for x in price:
    original_price = x.select_one('.discount_original_price')
    if original_price:
        # 원래 가격이 있을 경우
        price_value = original_price.text.replace('₩','').replace(',','').strip()
    else:
        # 원래 가격이 없으면 할인된 가격을 사용
        final_price = x.select_one('.discount_final_price')
        price_value = final_price.text.replace('₩','').replace(',','').strip()
        if price_value == 'Free':
            price_value = 0
    
    print(int(price_value))
```

```python
# 각 게임의 링크 추출
url_list = [link.attrs['href'] for link in soup.select('a.search_result_row')]
url_list
```
